````python
import os
import subprocess

# Set environment variable for FSL output type
os.environ["FSLOUTPUTTYPE"] = "NIFTI_GZ"
````

## Mass case

````python
root = '/workspace/data/VinDr-Mammo'
mass_root = os.path.join(root, "Mass", 'image')
result_root = os.path.join(root, "Mass", 'registered_oppositeSide_image')
os.makedirs(result_root, exist_ok=True)
tmp_root = os.path.join(root, "tmp-mass")
os.makedirs(tmp_root, exist_ok=True)
mass_image_list = os.listdir(mass_root)
print(f"Number of mass images: {len(mass_image_list)}")
````

````python
script_path = "/workspace/scripts/VinDr-Mammo_Registration_Mass.sh"

# Iterate over files and run the shell script
for filename in mass_image_list:
    print(f"[INFO] Running registration for {filename}")
    try:
        subprocess.run(
            ["bash", script_path, filename],
            check=True,
            env=os.environ  # pass updated environment
        )
    except subprocess.CalledProcessError as e:
        print(f"[ERROR] Failed on {filename}: {e}")
````

````python
result_image_list = os.listdir(result_root)
print(f"Number of registered images: {len(result_image_list)}")
````
